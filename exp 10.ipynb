```python
# S. Charan
def hanoi(n, source, auxiliary, target):
    if n == 1:
        print(source, "->", target)
        return
    hanoi(n - 1, source, target, auxiliary)
    print(source, "->", target)
    hanoi(n - 1, auxiliary, source, target)

hanoi(3, 'A', 'B', 'C')
```

```python
# S. Charan
a = [12, 11, 13, 5, 6]

for i in range(1, len(a)):
    key = a[i]
    j = i - 1
    while j >= 0 and a[j] > key:
        a[j + 1] = a[j]
        j -= 1
    a[j + 1] = key

print(a)
```
