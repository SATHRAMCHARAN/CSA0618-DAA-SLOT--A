```python
# S. Charan - Factorial (Iterative vs Recursive)

n = int(input("Enter n: "))

fact = 1
for i in range(1, n + 1):
    fact *= i
print("Iterative Factorial:", fact)

def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)

print("Recursive Factorial:", factorial(n))
```
