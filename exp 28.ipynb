```python
# Name: S.Charan
from itertools import permutations
from math import sqrt

def distance(city1, city2):
    return sqrt((city1[0] - city2[0])**2 +
                (city1[1] - city2[1])**2)

def tsp(cities):
    if not cities:
        return 0, []

    if len(cities) == 1:
        return 0, [cities[0], cities[0]]

    start = cities[0]
    min_distance = float("inf")
    best_path = []

    for perm in permutations(cities[1:]):
        path = [start] + list(perm) + [start]

        total = sum(
            distance(path[i], path[i + 1])
            for i in range(len(path) - 1)
        )

        if total < min_distance:
            min_distance = total
            best_path = path

    return min_distance, best_path

print("S.Charan - Q28")

tests = [
    [(1, 2), (4, 5), (7, 1), (3, 6)],
    [(2, 4), (8, 1), (1, 7), (6, 3), (5, 9)]
]

for i, cities in enumerate(tests, 1):
    d, path = tsp(cities)
    print("Test Case:", i)
    print("Shortest Distance:", d)
    print("Shortest Path:", path)
```
