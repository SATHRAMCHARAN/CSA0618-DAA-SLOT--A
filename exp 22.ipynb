```python
# Name: S.Charan
def find_peak_element(nums):
    left, right = 0, len(nums) - 1

    while left < right:
        mid = (left + right) // 2

        if nums[mid] < nums[mid + 1]:
            left = mid + 1
        else:
            right = mid

    return left

print("S.Charan - Q22")

for nums in [[1, 2, 3, 1], [1, 2, 1, 3, 5, 6, 4]]:
    index = find_peak_element(nums)
    print("Input:", nums)
    print("Peak Index:", index)
    print("Peak Value:", nums[index])
```
