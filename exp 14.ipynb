```python
# Name: S. Charan
def power(x, n):
    if n == 0:
        return 1
    half = power(x, n // 2)
    if n % 2 == 0:
        return half * half
    return x * half * half

print("Name: S. Charan")
print("Output:", power(2, 10))
```
