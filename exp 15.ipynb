```python
# Name: S. Charan
import math

points = [(1, 2), (4, 5), (7, 8), (3, 1)]
best_pair = None
min_dist = float('inf')

for i in range(len(points)):
    for j in range(i + 1, len(points)):
        d = math.dist(points[i], points[j])
        if d < min_dist:
            min_dist = d
            best_pair = (points[i], points[j])

print("Name: S. Charan")
print("Closest pair:", best_pair)
print("Distance:", round(min_dist, 3))
```
