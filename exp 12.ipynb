```python
# S. Charan
def count_inversions(a):
    count = 0
    for i in range(len(a)):
        for j in range(i + 1, len(a)):
            if a[i] > a[j]:
                count += 1
    return count

print(count_inversions([2, 4, 1, 3, 5]))
```
