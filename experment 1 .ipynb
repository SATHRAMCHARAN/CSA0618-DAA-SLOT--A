```python
name = "S. Charan"

def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

arr = [10, 25, 30, 45, 50]
key = 30

print("Name:", name)
print("Array:", arr)
print("Key:", key)

index = linear_search(arr, key)

if index != -1:
    print("Key found at index", index)
else:
    print("Key not found")
```
