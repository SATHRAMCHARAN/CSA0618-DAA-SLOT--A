```python
# Name: S.Charan
from itertools import permutations

def total_cost(assignment, cost_matrix):
    return sum(cost_matrix[i][assignment[i]]
               for i in range(len(assignment)))

def assignment_problem(cost_matrix):
    n = len(cost_matrix)
    best_assignment = None
    min_cost = float("inf")

    for assignment in permutations(range(n)):
        cost = total_cost(assignment, cost_matrix)

        if cost < min_cost:
            min_cost = cost
            best_assignment = assignment

    pairs = [(i + 1, task + 1)
             for i, task in enumerate(best_assignment)]

    return pairs, min_cost

print("S.Charan - Q29")

tests = [
    [[3, 10, 7], [8, 5, 12], [4, 6, 9]],
    [[15, 9, 4], [8, 7, 18], [6, 12, 11]]
]

for i, matrix in enumerate(tests, 1):
    assignment, cost = assignment_problem(matrix)
    print("Test Case:", i)
    print("Optimal Assignment:", assignment)
    print("Total Cost:", cost)
```
