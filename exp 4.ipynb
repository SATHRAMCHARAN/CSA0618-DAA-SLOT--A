```python
# Experiment 4: Fibonacci Series
# Name: S. Charan

n = int(input("Enter n: "))

a, b = 0, 1
fib = []

for i in range(n):
    fib.append(a)
    a, b = b, a + b

print("Fibonacci Series:", fib)
```
